```python
num_col=df.select_dtypes(include=['int','float']).columns
print(num_col)
```

```python
cat_col=df.select_dtypes(include=['object']).columns
print(cat_col)
```

```python
##SUBPLOT --HISTIOGRAM
plt.figure(figsize=(10,7))
for i in range(len(num_col)):
    plt.subplot(2,4,i+1)
    sns.histplot(x=df[num_col[i]])
    plt.tight_layout()
    
```

```python
plt.figure(figsize=(10,7))
for i in range(len(num_col)):
    plt.subplot(3,3,i+1)
    sns.histplot(x=df[num_col[i]])
    plt.tight_layout()
```

```python
###DENSITY COLUMN
plt.figure(figsize=(10,7))
for i in range(len(num_col)):
    plt.subplot(2,4,i+1)
    sns.kdeplot(x=df[num_col[i]])
    plt.tight_layout()
```

```python
###COUNTPLOT
plt.figure(figsize=(10,6))
for i in range(len(cat_col)):
    plt.subplot(2,2,i+1)
    sns.histplot(x=df[cat_col[i]])
    plt.tight_layout()
```

```python
plt.figure(figsize=(10,6))
for i in range(len(cat_col)):
    plt.subplot(2,2,i+1)
    sns.countplot(x=df[cat_col[i]])
    plt.tight_layout()
```
